# Laboratorio SP · simulador interactivo de anomalías

Hay uno o varios **cuerpos escondidos** que generan datos de potencial espontáneo
"observados", con ruido. El trabajo es reproducirlos: se ponen cuerpos, se mueven sus
parámetros y se miran las **métricas de error** hasta que el modelo explique los datos.

El modelo directo (Yüngül, 1950; Biswas, 2017), sumado para varios cuerpos por
superposición:

$$V(x) = \sum_i k_i\,\frac{(x-x_{0,i})\cos\theta_i + h_i\sin\theta_i}{\left[(x-x_{0,i})^2 + h_i^2\right]^{q_i}}$$

| Parámetro | Qué es | Qué controla |
|---|---|---|
| **q** | forma: 0.5 cilindro vertical · 1.0 cilindro horizontal · 1.5 esfera | ancho y colas |
| **k** | amplitud de polarización (mV·m^(2q−1)) | escala y signo |
| **h** | profundidad (m): al techo, al eje o al centro según la forma | amplitud **y** ancho |
| **θ** | ángulo de polarización (°) | simetría |
| **x₀** | posición horizontal (m) | solo desplaza |

## Cómo jugar

1. Elige cuántos **cuerpos ocultos** quieres (1 a 4) y si quieres ver sus formas como pista.
2. Mira la forma de los datos: ¿un pico, un escalón o una curva antisimétrica? Eso dice la
   **forma** y el **θ**. Agrega al modelo tantos cuerpos como creas que hay.
3. Mueve **x₀** hasta la anomalía y ajusta **h** con el **ancho** (más profundo = más ancho).
4. **Optimizar k:** como V es lineal en k, el mejor k sale directo por mínimos cuadrados.
5. **Inversión automática:** ajusta x₀, h, θ y k de todos los cuerpos desde varios
   **puntos de partida**. Si no todos llegan a la misma solución, hay mínimos locales.
6. **Probar formas:** invierte con todas las combinaciones de q y compara el χ².
7. Compara con **Revelar solución**.

## Cómo saber si confiar en un resultado

| Señal | Qué dice |
|---|---|
| **χ² reducido ≈ 1** | El modelo explica los datos hasta el nivel del ruido. Se calcula con N − p grados de libertad (p = 4 parámetros por cuerpo). Diferencias menores que el margen indicado (≈ 0.4 con 61 datos) son azar. |
| **χ² ≪ 1** | Se está ajustando ruido: el modelo tiene más cuerpos o parámetros de los que los datos justifican. |
| **Residuos sin patrón** | Si los residuos rojos se agrupan en una zona, falta un cuerpo o la forma es otra. |
| **± de cada parámetro** | Incertidumbre de 1σ. Dos soluciones cuyos valores se solapan dentro de ± no son distintas. Su honestidad se verifica en `SP_verificacion.ipynb`. |
| **Puntos de partida coincidentes** | Si la mayoría llega a la misma solución, es estable. Si no, hay mínimos locales. |
| **Parámetros acoplados (\|r\| > 0.9)** | Los datos no los separan: típicamente **k y h** (un cuerpo más profundo y más polarizado da casi la misma curva). |
| **Probar formas: varias «compatibles»** | Los datos no permiten decidir la forma. |
